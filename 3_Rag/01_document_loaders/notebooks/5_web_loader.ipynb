# Recursive URL
RecursiveUrlLoader is a LangChain tool that starts at a main webpage, automatically follows its child and sub-child links up to a set depth (max_depth), and extracts the text from every visited page into Document objects for your AI.

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
# create the loader

base_url = "https://docs.langchain.com/oss/python/integrations/document_loaders"

loader = RecursiveUrlLoader(url=base_url,
                            max_depth=2)
```

```python
# load the documents

documents = loader.load()
```

```python
# number of documents

len(documents)
```

```python
documents[0].metadata
```

```python
# metadat for first 10 documents
for i in range(10):
    doc = documents[i]
    print(doc.metadata, end="\n\n")
```
