```python
folder_path = r"E:\GenAI Class\1_EncodingAndEmbeddig\data"
story = []
for filename in os.listdir(folder_path):
    file_path = os.path.join(folder_path, filename)
    with open(file_path, encoding='unicode_escape') as f:
        corpus = f.read()
    raw_sent = sent_tokenize(corpus)
    for sent in raw_sent:
        story.append(simple_preprocess(sent))
```

```python
# min_count means that word minimum time occure atlest 5
# These all the value are the hyperparameter
# sg=0 mean CBOW and sg=1 means Skip-gram and by deafultvalue is 0
custom_model =gensim.models.Word2Vec(window=10, 
                                     min_count=1,
                                     vector_size=150)
```

```python
#  build vocabulary out of the data
custom_model.build_vocab(story)
```

```python
# custom_model.corpus_count- it gives the total no of sentece 
custom_model.train(story, total_examples=custom_model.corpus_count, epochs=10)
```
