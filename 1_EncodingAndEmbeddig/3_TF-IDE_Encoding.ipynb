```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = ["people watch movie", 
             "people watch cricket", 
             "people like movie", 
             "people like cricket"]
```

```python
tf_idf = TfidfVectorizer()
```

```python
df_idf_vector = tf_idf.fit_transform(documents)
```

```python
df_idf_vector
```

```python
df_idf_vector.toarray()
```
