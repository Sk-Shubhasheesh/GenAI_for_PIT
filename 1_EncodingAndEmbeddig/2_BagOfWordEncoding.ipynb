```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = ["people watch movie", 
             "people watch cricket", 
             "people like movie", 
             "people like cricket"]
```

AIM: Convert:Data:Number

https://scikit-learn.org/0.15/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html

```python
bow = CountVectorizer()
```

```python
# data preprocessing
bow_matrix = bow.fit_transform(documents)
```

```python
bow_matrix
```

```python
print("BOW MAtrix:\n", bow_matrix.toarray())
```

```python
# Get Vocabulary
print(bow.get_feature_names_out())
```
